## 12. Changing data type of object typed columns

```python
df['LoanID'] = df['LoanID'].astype('string')
df['Education'] = df['Education'].astype('string')
df['EmploymentType'] = df['EmploymentType'].astype('string')
df['MaritalStatus'] = df['MaritalStatus'].astype('string')
df["HasMortgage"] = df["HasMortgage"].map({
    "Yes": 1, 
    "No": 0
}).astype("int64")
df["HasDependents"] = df["HasDependents"].map({
    "Yes": 1, 
    "No": 0
}).astype("int64")
df['LoanPurpose'] = df['LoanPurpose'].astype('string')
df["HasCoSigner"] = df["HasCoSigner"].map({
    "Yes": 1, 
    "No": 0
}).astype("int64")
df['Education'] = df['Education'].map({
    "High School" : 0,
    "Bachelor's" : 1,
    "Master's" : 2,
    "PhD" : 3
})
df.info()
```
